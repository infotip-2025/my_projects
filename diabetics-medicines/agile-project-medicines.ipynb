## User stories

- As a patient I want bo be able to check **what is the last day with full dose secured** of a particular medicine so that I can **safely order them in time**.

- As a patient I want to know **when I collected particular medicine** including its quantity and dose so that I can trace how my **prescribed medicines list changed over time**.
 
- As a ptient I want to know **what were my particular medicine supplies** including dose split so that I can **trace how my supplies change over time**.
