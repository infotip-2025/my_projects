Forxiga (dapagliflozyna) to lek stosowany w leczeniu cukrzycy typu 2, niewydolności serca i przewlekłej choroby nerek. Chociaż jest ogólnie dobrze tolerowana, jak wszystkie leki, może mieć skutki uboczne, z których niektóre są poważne.

Ważne jest, aby zrozumieć, że "szanse" na poważne skutki uboczne są generalnie **niskie**, ale nie wynoszą zero. Oto zestawienie niektórych poważnych skutków ubocznych i ich przybliżonej częstości na podstawie badań klinicznych i raportów po wprowadzeniu leku do obrotu:

**Poważne skutki uboczne (rzadkie do bardzo rzadkich):**

* **Kwasica ketonowa (DKA):** Jest to poważny i zagrażający życiu stan, w którym organizm wytwarza wysokie poziomy kwasów we krwi, zwanych ketonami.
    * **Szanse:** Szacuje się, że występuje u **od 1 na 1000 do 1 na 10 000 pacjentów** leczonych inhibitorem SGLT2, takim jak dapagliflozyna. Chociaż rzadkie, jest to poważne zmartwienie, zwłaszcza że może wystąpić nawet przy prawidłowym lub tylko nieznacznie podwyższonym poziomie cukru we krwi.
    * **Czynniki ryzyka:** Zwiększone ryzyko, jeśli masz cukrzycę typu 1, historię DKA, jesteś chory (gorączka, wymioty, biegunka), stosujesz dietę bardzo niskowęglowodanową, pościsz lub masz znaczne zmniejszenie dawki insuliny.
    * **Objawy:** Nudności, wymioty, ból brzucha, zmęczenie, trudności w oddychaniu, dezorientacja, nadmierne pragnienie, słodki zapach oddechu. **W przypadku wystąpienia tych objawów należy natychmiast zasięgnąć pomocy medycznej.**

* **Zgorzel Fourniera (martwicze zapalenie powięzi krocza):** Jest to bardzo rzadka, ale poważna i szybko postępująca infekcja bakteryjna tkanek wokół narządów płciowych i odbytu. Może prowadzić do hospitalizacji, operacji, a nawet śmierci.
    * **Szanse:** Szacuje się, że występuje u około **1 na 100 000 pacjentów** leczonych inhibitorem SGLT2. Chociaż niezwykle rzadkie, wydaje się być około 12 razy częstsze u pacjentów leczonych inhibitorami SGLT2 niż innymi lekami przeciwcukrzycowymi. Większość przypadków występuje u mężczyzn.
    * **Objawy:** Silny ból, tkliwość, zaczerwienienie lub obrzęk w okolicy pachwiny lub krocza, któremu towarzyszy gorączka lub złe samopoczucie. **Natychmiast zasięgnąć pomocy medycznej.**

* **Poważne infekcje dróg moczowych (ZUM), w tym posocznica moczowa i odmiedniczkowe zapalenie nerek:** Chociaż ZUM są częstym skutkiem ubocznym dapagliflozyny, ciężkie ZUM wymagające hospitalizacji lub mogące prowadzić do infekcji nerek (odmiedniczkowe zapalenie nerek) lub ogólnoustrojowej infekcji (posocznica moczowa) są rzadsze.
    * **Szanse:** Mniej powszechne niż ogólne ZUM. W badaniach klinicznych ogólne ZUM występowały u około 6% osób, ale poważne ZUM są rzadsze.

* **Odwodnienie i niskie ciśnienie krwi (hipotensja):** Dapagliflozyna działa poprzez usuwanie nadmiaru cukru i wody z organizmu z moczem. Może to prowadzić do odwodnienia, zwłaszcza u osób starszych, przyjmujących leki moczopędne lub z problemami z nerkami. Ciężkie odwodnienie może prowadzić do niskiego ciśnienia krwi, zawrotów głowy, omdleń i potencjalnie ostrego uszkodzenia nerek.
    * **Szanse:** Działania niepożądane związane ze zmniejszoną objętością wewnątrznaczyniową (w tym odwodnienie i hipotensja) wystąpiły u **0,1% do 1%** w niektórych badaniach.

* **Ciężkie reakcje alergiczne (anafilaksja):** Podobnie jak w przypadku większości leków, ciężka reakcja alergiczna jest możliwa, ale rzadka.
    * **Szanse:** W badaniach klinicznych 0,3% osób miało reakcję alergiczną. Ciężka reakcja (anafilaksja) jest jeszcze rzadsza.
    * **Objawy:** Obrzęk twarzy, warg, języka lub gardła, trudności w oddychaniu, ciężka wysypka. **Natychmiast wezwij służby ratunkowe, jeśli wystąpią te objawy.**

* **Niski poziom cukru we krwi (hipoglikemia):** Chociaż dapagliflozyna zazwyczaj nie powoduje niskiego poziomu cukru we krwi sama w sobie, ryzyko znacznie wzrasta, gdy jest przyjmowana z innymi lekami przeciwcukrzycowymi, takimi jak insulina lub pochodne sulfonylomocznika.

**Ogólne informacje o częstości występowania skutków ubocznych:**

* **Bardzo często (dotyczy więcej niż 1 na 10 osób):** Hipoglikemia (w przypadku stosowania z insuliną lub pochodnymi sulfonylomocznika), zwiększony hematokryt (liczba czerwonych krwinek).
* **Często (dotyczy od 1 na 10 do 1 na 100 osób):** Grzybicze infekcje narządów płciowych (częściej u kobiet), infekcje dróg moczowych, zwiększone oddawanie moczu, ból pleców, zawroty głowy, łagodna wysypka skórna, zmiany poziomu cholesterolu/tłuszczów.

* **Rzadko (dotyczy od 1 na 1000 do 1 na 10 000 osób):** Kwasica ketonowa.
* **Bardzo rzadko (dotyczy mniej niż 1 na 10 000 osób):** Zgorzel Fourniera, ostre zakażenie nerek (posocznica moczowa, odmiedniczkowe zapalenie nerek), zapalenie trzustki.
